```python
import pandas as pd
import numpy as np
```

```python


df=pd.DataFrame({
                 'Name': ['Alice', 'Charlie', 'Bob','Alice'],
                 'Age' : [25, np.nan , 35,25],
                 'Salary' : [50000, 60000, np.nan,50000],
})
```

```python
df
```

#Duplicate Data Remove hojay ga

```python
df=df.drop_duplicates()
```

```python
df
```

```python
#Missing Values ko remove krny k liye
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Salary'].fillna(df['Salary'].mean(),inplace=True)
```
